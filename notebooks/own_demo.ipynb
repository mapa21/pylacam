```python
import numpy as np
from pycam import (
    LaCAM,
    Grid,
    Config,
)
```

```python
ROWS = 5
COLS = 7
grid : Grid = np.ones((2, ROWS, COLS), dtype=bool)
starts, goals = Config(), Config()
starts.init_populate([(0, 0, 4), (0, 1, 6), (0, 0, 2), (0, 4, 4)])
goals.init_populate([(0, 1, 5), (0, 1, 5), (0, 1, 1), (0, 1, 1)])
planner = LaCAM()
```

```python
timeout = np.iinfo(np.int32).max
solution = planner.solve(grid, starts, goals, time_limit_ms=timeout, verbose=2)
```

```python
for pos in solution:
    print(pos.positions)
```
